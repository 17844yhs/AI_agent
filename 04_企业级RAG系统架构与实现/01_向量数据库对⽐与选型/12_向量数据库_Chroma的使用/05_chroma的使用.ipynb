```python
# uv add chromadb
import chromadb
```

```python
# 创建客户端 (持久化硬盘)
client = chromadb.PersistentClient(path='./chroma_db')
```

```python
# 创建集合
collection = client.create_collection(
    name ='document_collection',
    metadata={'hnsw:space':'l2'}  # 使用L2空间
)
```

```python
# 准备数据
documents = [
    "向量数据库是AI应用的核心基础设施",
    "RAG系统需要高效的向量检索能力",
    "相似度搜索是向量数据库的核心功能"
]
ids = ["doc1", "doc2", "doc3"]
metadatas = [
    {"category": "技术", "source": "文档1"},
    {"category": "技术", "source": "文档2"},
    {"category": "技术", "source": "文档3"}
]
```

```python
# 添加文档
collection.add(
    documents=documents,
    ids = ids,
    metadatas = metadatas
)
```

```python
# 查询
results = collection.query(
    query_texts= ["向量数据库的核心功能"],
    n_results=2,
    where = {'category':'技术'}    # 过滤条件
)
results
```

```python
print("Chroma查询结果:")
for i, doc in enumerate(results["documents"][0]):
    print(f"{i+1}. {doc}")
    print(f"   距离: {results['distances'][0][i]:.4f}\n")
```
