```python
import os
from dotenv import load_dotenv

load_dotenv()

from langchain_community.llms import Tongyi

# 获取环境中KEY
tongyi_key = os.environ.get('QWEN_KEY')
# 设置KEY
os.environ["DASHSCOPE_API_KEY"] = tongyi_key
# 创建模型
llm = Tongyi()
```

### LCEL重写传统LLMChain

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser


prompt = ChatPromptTemplate.from_template('为{product}写一个吸引人的广告语.')

# 最简单的方式
chain = prompt | llm | StrOutputParser()
# 调用
rs = chain.invoke({'product':'智能手表'})
print(rs)
```

```python
from langchain_core.runnables import RunnablePassthrough,RunnableLambda
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = ChatPromptTemplate.from_template('为{product}写一个吸引人的广告语.')

def pre_process(product_input):
    return product_input['product'].upper()

chain = (
    RunnablePassthrough.assign(product_name=RunnableLambda(pre_process))
    | RunnablePassthrough.assign(ad= prompt | llm | StrOutputParser())
)

rs = chain.invoke({'product':'apple watch'})
print(f'广告产品:{rs.get("product_name")}')
print(f'广告语:{rs.get("ad")}')
```

### LCEL重写SimpleSequentialChain

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser


prompt1 = ChatPromptTemplate.from_template("为一部关于{topic}的电影起一个响亮的标题。")
prompt2 = ChatPromptTemplate.from_template("为标题为'{title}'的电影写一段简短的剧情简介。")

chain = (
    {'title': prompt1 | llm |StrOutputParser()}
    | prompt2 |llm | StrOutputParser()
)

# 调用
rs = chain.invoke({'topic': '人工智能觉醒'})
print(rs)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough, RunnableLambda

prompt1 = ChatPromptTemplate.from_template("基予{topic}的电影起一个响亮的标题。")
prompt2 = ChatPromptTemplate.from_template("为标题为'{title}'的电影写一段简短的剧情简介。")

# 步骤1： 生成标题
title_genrator = prompt1 | llm | StrOutputParser()

# 步骤2： 获取电影简介
chain = (
    RunnablePassthrough.assign(title=title_genrator)
    | RunnablePassthrough.assign(summary = prompt2 |llm | StrOutputParser())
)

rs = chain.invoke({"topic": "人工智能觉醒"})
print(rs)
```

### LCEL重写SequentialChain

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# 定义 prompts
prompt1 = ChatPromptTemplate.from_template("用{language}写一段关于{era}时代的评语。")
prompt2 = ChatPromptTemplate.from_template("将以下评语翻译成中文：\n{review}")

# 创建链
chain = (
    {'review':prompt1|llm|StrOutputParser()}
    | prompt2 | llm | StrOutputParser()
)
rs = chain.invoke({'language':'英语', 'era':'维多利亚'})
print(rs)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

# 定义 prompts
prompt1 = ChatPromptTemplate.from_template("用{language}写一段关于{era}时代的评语。")
prompt2 = ChatPromptTemplate.from_template("将以下评语翻译成中文：\n{review}")

# 步骤1
generator = prompt1 | llm | StrOutputParser()

# 步骤2： 获取电影简介
chain = (
    RunnablePassthrough.assign(review=generator)
    | RunnablePassthrough.assign(result = prompt2 |llm | StrOutputParser())
)

rs = chain.invoke({'language':'英语', 'era':'维多利亚'})
print(rs)
```

### LCEL实现RouterChain

```python
# 方法1：使用RunnableLambda（不推荐）
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableLambda

prompt1 = ChatPromptTemplate.from_template("你是一位物理教授。回答这个问题：{input}")
prompt2 = ChatPromptTemplate.from_template("你是一位数学教授。回答这个问题：{input}")

physics_chain = prompt1 |llm | StrOutputParser()  # 创建一个物理链
math_chain = prompt2 |llm | StrOutputParser()  # 创建一个数学链

# 路由的逻辑
def router(prompt:str):
    if '物理' in prompt['input'] or '牛顿' in prompt['input'] or 'physics' in prompt['input'].lower():
        return physics_chain
    else:
        return math_chain
# 构建chain
chain = RunnableLambda(router)

rs = chain.invoke({"input": "牛顿在物理上如何解释牛顿第二定律？"})
print(rs)
```

```python
# 方法二：使用RunnableBranch(推荐)
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableBranch


prompt1 = ChatPromptTemplate.from_template("处理物理问题：{input}")
prompt2 = ChatPromptTemplate.from_template("处理数学问题：{input}")
prompt3 = ChatPromptTemplate.from_template("通用问题：{input}")

physics_chain = prompt1 |llm | StrOutputParser()  # 创建一个物理链
math_chain = prompt2 |llm | StrOutputParser()  # 创建一个数学链
default_chain = prompt3 |llm | StrOutputParser()

# 使用RunnableBranch 创建一个路由链
branch = RunnableBranch(
    (lambda x:"物理" in x["input"], physics_chain), 
    (lambda x:"math" in x["input"],math_chain), 
    default_chain)

rs = branch.invoke({"input": "牛顿在物理上如何解释牛顿第二定律？"})
rs
```
