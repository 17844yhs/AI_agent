### 对话提示词模版

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.messages import SystemMessage,HumanMessage

# 定义模板
sys_template = "你是数学老师,请以{style}的风格回答问题"
user_template = "请用简单易懂的方式解释：{question}"

# 创建对话模板-构造方法
prompt_template1 = ChatPromptTemplate([
    ('system',sys_template),
    ('human',user_template)
])

# 创建对话模板-from_messages
prompt_template2 = ChatPromptTemplate.from_messages([
    ('system',sys_template),
    ('human',user_template)
])

# 创建对话模板
prompt_template3 = ChatPromptTemplate.from_messages([
    SystemMessage(sys_template),
    HumanMessage(user_template)
])

print(prompt_template1)
print(prompt_template2)
print(prompt_template3)
```

```python
# 填充变量
prompt1 = prompt_template1.invoke({'style':'通俗易懂','question':'勾股定理是什么？'})
prompt2 = prompt_template1.format(style='通俗易懂',question='勾股定理是什么？')
prompt3 = prompt_template1.format_messages(style='通俗易懂',question='勾股定理是什么？')
prompt4 = prompt_template1.format_prompt(style='通俗易懂',question='勾股定理是什么？')

print(prompt1)
print('-'*50)
print(prompt2)
print('-'*50)
print(prompt3)
print('-'*50)
print(prompt4)
```

### 少量样本提示词模板

```python
from langchain_community.llms import Tongyi

# 获取环境中KEY
tongyi_key = os.environ.get('QWEN_KEY')
# 设置KEY
os.environ["DASHSCOPE_API_KEY"] = tongyi_key

# 创建模型
llm = Tongyi()
```

```python
# Zero-shot 失败的情况
print("=== Zero-shot 失败 ===")
question = "2 ✿ 3 是多少？"
res_zero = llm.invoke(question)
print(f"问题：{question}")
print(f"回答：{res_zero}")  # 无法理解 ✿ 的含义
```

```python
# Few-shot 解决方案
# print("\n=== Few-shot 解决 ===")

# 定义示例
examples = [
    {"input": "2 ✿ 3", "output": "6 (因为 ✿表示乘法)"},
    {"input": "3 ✿ 7", "output": "21 (因为 ✿ 表示乘法)"},
]

# 构建 Few-shot 提示
from langchain_core.prompts import FewShotPromptTemplate, PromptTemplate

# 普通提示词模版
example_prompt = PromptTemplate.from_template("输入：{input}\n输出：{output}")
# 少量样本提示词模版
prompt = FewShotPromptTemplate(
    examples=examples,
    example_prompt=example_prompt,
    suffix="输入：{input}\n输出：",
    input_variables=["input"]
)

# 格式化提示
few_shot_prompt = prompt.format(input="6 ✿ 6")

res_few = llm.invoke(few_shot_prompt)

# print(f"Few-shot 提示：\n{few_shot_prompt}")
print(f"\nFew-shot 回答：{res_few}")  # 36 (因为 ✿ 表示乘法)
```

```python
# 1. 准备示例（城市提取场景）
examples = [
    {"input": "北京天气怎么样", "output": "北京市"},
    {"input": "南京下雨吗", "output": "南京市"},
    {"input": "武汉热吗", "output": "武汉市"}
]

# 2. 创建示例模板
example_prompt = PromptTemplate.from_template(
    "Input: {input}\nOutput: {output}"
)

# 3. 构建 FewShot 提示
few_shot_prompt = FewShotPromptTemplate(
    examples=examples,
    example_prompt=example_prompt,
    suffix="Input: {city_input}\nOutput:",
    input_variables=["city_input"]
)

# 4. 格式化和调用
prompt_text = few_shot_prompt.format(city_input="长沙多少度")

print("=== 生成的 Prompt ===")
print(prompt_text)

# 5. 调用模型
info = llm.invoke(prompt_text)
print(info)
```
