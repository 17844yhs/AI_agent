# 📖 介绍

**LCEL (LangChain Expression Language)** 是 LangChain 推出的**声明式链式调用语法**。它基于 **Runnable 协议**，提供了一套统一的标准来构建异步、流式和并行的 AI 应用。

---

# 💡 为什么要使用 LCEL

| 特性 | 旧版 Chain | LCEL |
|------|-----------|------|
| **语法风格** | 命令式，需要显式定义输入输出键 | 声明式，使用管道操作符 `\|` |
| **异步支持** | 需单独编写异步代码 | 原生支持 `async/await`，一键切换 `.ainvoke` |
| **流式输出** | 配置复杂 | 内置支持，`.stream()` 开箱即用 |
| **并行执行** | 难以实现 | `RunnableParallel` 自动优化并行调用 |
| **传入/传出** | 需手动管理字典数据 | 自动透传中间数据 |

```python
import os
from dotenv import load_dotenv

load_dotenv()

from langchain_community.llms import Tongyi

# 获取环境中KEY
tongyi_key = os.environ.get('QWEN_KEY')
# 设置KEY
os.environ["DASHSCOPE_API_KEY"] = tongyi_key
# 创建模型
llm = Tongyi()
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
# ❌ 传统方式：手动分步调用
prompt = ChatPromptTemplate.from_template("告诉我一个关于{topic}的笑话")
output_parser = StrOutputParser()

# 第一步：prompt
format_prompt = prompt.invoke({'topic': '程序员'})
# 第二步：LLM调用
llm_output = llm.invoke(format_prompt)
# 第三步：格式化输出
result = output_parser.parse(llm_output)
print(result)
```

```python
# ✅ LCEL方式：一行搞定
chain = prompt | llm | output_parser

# 调用链
result = chain.invoke({'topic': '程序员'})
print(result)
```
