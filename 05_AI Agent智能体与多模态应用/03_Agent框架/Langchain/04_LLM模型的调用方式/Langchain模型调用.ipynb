### 调用模型的方式

```python
import os
from dotenv import load_dotenv

load_dotenv()
```

#### 非对话模型
每次调用独立处理输入，无法自动关联历史对话（需手动拼接历史文本）

```python
# 非对话模式
from langchain_community.chat_models import ChatZhipuAI

# 获取环境中KEY
zhipu_key = os.environ.get('ZHIPU_KEY')

# 设置KEY
os.environ["ZHIPUAI_API_KEY"] = zhipu_key

# 创建模型
llm = ChatZhipuAI(model="glm-4.6v-flash")
# 使用模型
info = llm.invoke("夏天适合吃哪个水果？回复几个水果名称就好")
info
```

#### 对话模型

通过消息列表维护上下文，模型可基于完整对话历史生成回复。

#### 消息类型

| 场景 | 推荐消息类型 | 备注 |
|:---|:---|:---|
| 设定角色、风格、背景 | `SystemMessage` | 作为对话的开始 |
| 用户提问、输入指令 | `HumanMessage` | 最常见的输入 |
| AI生成文本回复 | `AIMessage` | 标准的输出 |
| 工具执行返回结果 | `ToolMessage` | 通常自动创建，作为 tool_calls 的响应 |
| 需要非常特殊的角色 | `ChatMessage` | 需要显式指定 role 字符串 |
| 处理流式输出 | `AIMessageChunk` | 临时类型，最终会合并为 AIMessage |

```python
# 对话模型
from langchain_community.chat_models import ChatZhipuAI

# 获取环境中KEY
zhipu_key = os.environ.get('ZHIPU_KEY')

# 设置KEY
os.environ["ZHIPUAI_API_KEY"] = zhipu_key

# 创建模型
llm = ChatZhipuAI(model="glm-4.6v-flash")

# 创建提示词
messages = [
    ('system','你是个饮食健康专家，请根据用户的输入给出相应的建议。'),
    ('human','夏天吃什么水果合适?')
]

# 调用模型
info = llm.invoke(messages)
info
```

```python
from langchain_core.messages import SystemMessage,HumanMessage

messages = [
    SystemMessage(content='你是个饮食健康专家，请根据用户的输入给出相应的建议。'),
    HumanMessage(content='夏天吃什么水果合适?')
]
# 调用模型
info = llm.invoke(messages)
info
```
