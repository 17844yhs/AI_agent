```python
import json

class PromptManager:
    def __init__(self,config_path:str='prompts.json'):
        with open(config_path,'r',encoding='utf-8') as f:
            self.prompts = json.load(f)

    def get_prompt(self,category:str,key:str,**kwargs) -> str:
        prompt_data =  self.prompts.get(category).get(key)
        prompt = prompt_data.get('system')   # 获取提示词

        if "requirements" in prompt_data:
            prompt += '\n要求：'
            for i,req in enumerate(prompt_data.get('requirements'),1):
                prompt += f'\n{i}. {req}'

        # 变量的赋值
        if kwargs:
            for key,value in kwargs.items():
                prompt = prompt.replace(f'{{{{{key}}}}}',str(value))
                
        return prompt


manager = PromptManager()
prompt = manager.get_prompt(
    'customer_service',
    'welcome',
    company_name='ABC公司'
)
print(prompt)
```

```python
import yaml   # pip install pyyaml

class YamlPromptManager:
    def __init__(self,config_path:str='prompts.yaml'):
        with open(config_path,'r',encoding='utf-8') as f:
            self.prompts = yaml.safe_load(f)

    def get_prompt(self,category:str,key:str, **kwargs) ->str:
        prompt_data = self.prompts.get(category,{}).get(key,{})

        if not prompt_data:
            raise ValueError(f'找不到提示词：{category}.{key}')
        
        prompt = prompt_data.get('system','')  # 获取提示词

        # 添加要求列表
        if 'requirements' in prompt_data:
            prompt += '\n要求：'
            for i ,req in enumerate(prompt_data.get('requirements',[]),1):
                prompt += f'\n{i}. {req}'

        # 变量替换
        if kwargs:
            for var_key,var_value in kwargs.items():
                prompt = prompt.replace(f'{{{{{var_key}}}}}',str(var_value))

        return prompt
yaml_prompt = YamlPromptManager()
prompt = yaml_prompt.get_prompt(
    'customer_service',
    'welcome',
    company_name='ABC公司'
    )
print(prompt)
```
